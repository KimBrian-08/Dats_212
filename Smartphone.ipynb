Exercise 2: Smartphone Battery Level
Create a class named Smartphone with:

Attributes: brand and battery_level (default set to 100).
A method use_phone(minutes) that reduces battery_level by 1% for every 5 minutes used.
A method charge() that resets battery_level back to 100.
Test using the phone for 30 minutes, print the remaining battery, charge it, and print it again.

```python
from typing import Any
class Smartphone:
    def __init__(self,brand: Any| None, battery_level: int) -> None:
        self.brand = brand
        self.battery_level = battery_level
        
    def use_phone(self, minutes: int) -> list[int]:
        if self.battery_level <= 0:
            print("Battery low! Shutting down")
        return [ self.battery_level - (minutes // 5)]
    
    def charge(self) -> None:
        self.battery_level = 100
        print("Battery is fully charged!")
    
sample = Smartphone("iphone", 100)
remaining_perc = sample.use_phone(30)
print (remaining_perc)
```

Exercise 1: Student Introduction
Create a class named Student.

Add two instance variables:

name
course
Add an instance method named introduce() that prints the student's name and course.

Create two student objects and call introduce() for both.

```python
class Student:
    def __init__(self, name:str, course: str) -> None:
        self.name = name
        self.course = course
        
    def introduce(self) -> str:
        return f"Hello, my name is {self.name} and taking {self.course} course."

student1 = Student("Hamilton", "Computing and statistics")
print(student1.introduce())

student2 = Student("Charles", "Art")
print(student2.introduce())
```
