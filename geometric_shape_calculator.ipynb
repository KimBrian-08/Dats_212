Geometric Shape Calculator
Create a base class Shape with an area() method returning 0.
Create subclasses Rectangle (attributes: width, height) and Circle (attribute: radius).
Override area() in both subclasses to compute their respective mathematical areas.
Place different shapes into a list, iterate through them, and print each shape's area polymorphically.

```python
class Shape():
    #base class
    def area(self):
        return 0
    
class Rectangle(Shape):
    #subclass of Shape
    def __init__(self, width, height):
        self.width = width
        self.height = height
        
    def area(self):
        return self.width * self.height
    
class Circle(Shape):
    
    def __init__(self, radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * self.radius * self.radius
    
shapes = [Shape(), Rectangle(35, 30), Circle(15), Rectangle(75, 55)]

for shape in shapes:
    print(f"Area: {shape.area()}")
    

```

Comparing Bank Accounts with Operator Overloading

Create an Account class with attributes owner and balance.
Overload the greater-than operator (__gt__) to compare two account balances.
Instantiate two accounts and check which account has a higher balance using acc1 > acc2 in an if block.

```python
class Account:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance
    
    def __gt__(self, other):
        return self.balance > other.balance
    
acc1 = Account("Hamilton", 12000)
acc2 = Account("Marc", 11000)

if acc1 > acc2:
    print(f"{acc1.owner} has more money than {acc2.owner}")
    
else:
    print(f"{acc2.owner} has more money than {acc1.owner}")
        
    
```

Create a base class called Animal that requires every subclass to implement a make_sound() method.

Create three subclasses:

Dog
Cat
Cow
Each subclass should provide its own implementation of the make_sound() method.

Then:

Create an object of each class.
Call the make_sound() method for each object.
Display the appropriate sound for each animal.
Expected Output:

Dog says: Woof!
Cat says: Meow!
Cow says: Moo!
Challenge: Use Python's ABC and @abstractmethod to ensure that every concrete subclass must implement make_sound().

```python
from abc import ABC, abstractmethod

class Animal (ABC):
    
    @abstractmethod
    def make_sound(self):
        pass
    

class Dog(Animal):
    def make_sound(self):
        return "Woof!"


class Cat(Animal):
    def make_sound(self):
        return "Meow!"


class Cow(Animal):
    def make_sound(self):
        return "Moo!"
    
animals = [Dog(), Cat(), Cow()]

for animal in animals:
    print(f"{type(animal).__name__} says: {animal.make_sound()}")
```
