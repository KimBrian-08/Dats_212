Exercise 1: Secured Bank Account
Create a class BankAccount with public attribute account_holder and private attribute __balance.
Implement classic methods get_balance(), deposit(amount), and withdraw(amount).
Ensure withdraw() prevents users from overdrawing the balance.
Instantiate an account and demonstrate successful deposit and blocked overdrawing.

```python
class BankAccount:
    def __init__(self, account_holder: str, balance: float = 0.0) -> None:
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self) -> float:
        return self.__balance

    def deposit(self, amount: float) -> None:
        if amount > 0:
            self.__balance += amount
            print(f"Successfully deposited kshs {amount}. New balance: kshs {self.__balance}")
        else:
            print("Deposit amount must be greater than zero.")

    def withdraw(self, amount: float) -> None:
        if amount <= 0:
            print("Withdrawal amount must be greater than zero.")
        elif amount > self.__balance:
            print(f"Withdrawal failed: Insufficient funds. Requested kshs {amount}, available balance is kshs {self.__balance}.")
        else:
            self.__balance -= amount
            print(f"Successfully withdrew kshs {amount}. Remaining balance: kshs {self.__balance}")



account = BankAccount(account_holder="Hamilton", balance=1000.0)
print(f"Account Holder: {account.account_holder}")
print(f"Initial Balance: kshs{account.get_balance()}\n")

account.deposit(50.0)

account.withdraw(230.0)
account.withdraw(7000.0)

```

Exercise 2: Temperature Control with @property
Create a class Thermometer that takes private __celsius in __init__.
Use @property for celsius to read the temperature.
Add a @celsius.setter that prevents setting values below absolute zero (-273.15°C).
Test reading and updating with both valid and invalid values.

```python
class Thermometer:
    def __init__(self, celsius: float):
        self.celsius = celsius

    @property
    def celsius(self) -> float:
        # the getter method for celcius
        return self.__celsius

    @celsius.setter
    def celsius(self, value: float) -> None:
        #the setter methos for celcius that validates the temperature
        if value < -273.15:
            raise ValueError(
                f"Temperature cannot be below absolute zero (-273.15°C). Got: {value}°C"
            )
        self.__celsius = value



t = Thermometer(20.0)
print(f"Initial Temperature: {t.celsius}°C")

t.celsius = 36.6
print(f"Updated Temperature: {t.celsius}°C")

    
try:
    t.celsius = -500.0
except ValueError as e:
    print(f"Error: {e}")

print(f"Final Temperature: {t.celsius}°C")
```
