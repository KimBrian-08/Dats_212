### Assignment 1: Student Management System

#### Project Overview

```
This notebook implements a Student Management System designed to track student enrolment details within a university. The system leverages core Object-Oriented Programming principles in Python, including instance variables, instance methods, class variables, class methods, and static methods, to model student records and administrative functions effectively.

#### 1. Class Definition and OOP Implementation

The `Student` class is defined below. It incorporates the following components:
- Instance Variables: `name`, `student_id`, and `programme` to store specific student data.
- Class Variable: `university` to store a shared institutional name.
- Instance Methods: `display_details()` to output student information and `change_programme()` to update a student's enrolled course.
- Class Method: `change_university()` to modify the shared university name across all instances.
- Static Method: `is_valid_id()` to validate a student ID based on length criteria (minimum 5 characters).

````python
class Student:
    # Class variable shared by all instances
    university = "Kabarak University"

    def __init__(self, name, student_id, programme):
        # Instance variables
        self.name = name
        self.student_id = student_id
        self.programme = programme

    def display_details(self):
        
        """Instance method to display student details."""
        
        print(f"Student Name: {self.name}")
        print(f"Student ID: {self.student_id}")
        print(f"Programme: {self.programme}")
        print(f"University: {self.university}")
        print("-" * 67)
        print(" ")

    def change_programme(self, new_programme):
        
        """Instance method to change the student's programme."""
        
        print(f"Updating {self.name}'s programme from '{self.programme}' to '{new_programme}'.")
        self.programme = new_programme

    @classmethod
    def change_university(cls, new_university):
        
        """Class method to change the shared university name."""
        
        print(f"Changing university name from '{cls.university}' to '{new_university}'.")
        cls.university = new_university

    @staticmethod
    def is_valid_id(student_id):
        
        """Static method to validate the student ID.
           A valid student ID must contain at least 5 characters."""
           
        if len(str(student_id)) >= 5:
            return True
        else:
            return False
````

#### 2. Object Instantiation and Method Testing
The following code block demonstrates the creation of two `Student` objects and tests all defined methods. This verifies that:
1. Student details can be displayed correctly.
2. A student's programme can be updated.
3. Both students share the same class variable.
4. The shared university name can be modified globally.
5. The static method successfully validates student IDs.

````python
 # Two Student objects created to demonstrate the display_details() method.
 
print("------>>> Student Details <<<------")
print(" ")
student1 = Student("Itadori Yuji", "CS491785", "BSc Computer Science")
student2 = Student("Sung Jinwoo", "DS175390", "BSc Data Science")

student1.display_details()
student2.display_details()
print(" ")

# 2. Demonstrate the change_programme() instance method

print("\n------>> Updating Programme <<------")
student1.change_programme("BSc Data Science")
print("\n-->> Updated Student Details for Itadori Yuji <<--")
student1.display_details()

# 3. Demonstrate that both objects access the same class variable

print("\n--->> Demonstrating Shared Class Variable <<---")
print(f"student1 university: {student1.university}")
print(f"student2 university: {student2.university}")
print("Both students share the same university class variable.\n")
print(" ")

# 4. Demonstrate the change_university() class method
print("---------->> Updating University Name <<----------")
Student.change_university("Strathmore University")
print("\n--- Student Details After University Name Change ---")
student1.display_details()
student2.display_details()
print(" ")

# 5. Demonstrate the is_valid_id() static method
print("----->> Validating Student IDs <<-----")
test_id_1 = "CS491785"  # Valid (more than 5 characters)
test_id_2 = "A12"      # Invalid (less than 5 characters)

print(f"Is '{test_id_1}' a valid ID? {Student.is_valid_id(test_id_1)}")
print(f"Is '{test_id_2}' a valid ID? {Student.is_valid_id(test_id_2)}")
````
